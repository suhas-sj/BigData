
## 1. Bash Loops

### Bash while loop

A simple while loop is shown below:

```bash
while condition; do
  <command-list>;
done
```

The while loop executes **command-list** until **condition** no longer returns true.

When **condition** fails, the script continues with the command following **done**.

Condition can be any expression or command that returns a status.

```bash
#!/bin/bash
COUNTER=0
while [ $COUNTER -lt 10 ]; do
   echo "The counter is ${COUNTER}"
   let COUNTER=COUNTER+1
done
```

The above while loop prints a count from 0 to 9.


````python
!COUNTER=0; while [ $COUNTER -lt 10 ]; do echo "The counter is ${COUNTER}"; let COUNTER=COUNTER+1; done
````

### Bash for loop

```bash
for name in word;
do
   <list>;
done
```
The **for** loop above expands **word** into a list of items.

It replaces **name** with each item and performs **<list>** of commands in each iteration.

Using bash C-style **for** loop.

```bash
#!/bin/bash
for (( COUNT=1; COUNT <= 5; COUNT++ )) ;
do
    echo ${COUNT}
done
```

````python
#What does the above loop print?
!for (( COUNT=1; COUNT <= 5; COUNT++ )); do echo ${COUNT}; done
````

## 2. Functions

Functions in Bash are a way to group commands using a single name for the group.

**Functional decomposition** is decomposing a problem into high-level functions, decompose each of the high-level functions into lower-level functions, and so on.

### 2.1 Properties of Functions

**Abstraction**: We can focus on individual building blocks rather than the whole structure.

**Modularity**: Wrote a handy, generalized function? Use it in your other scripts!

**Readability**: Smaller code blocks are easier to wrap your mind around.

**Provability**: Decompose a problem into a set of provable functions. Prove the incorrectness of a function using tests.

### 2.2 Writing Functions Well

Keep functions small.

They should do one thing. They should do it well. They should do it only.

**Stepdown Rule:** The statements within a function are all at the same level of abstraction. Supports top-down narrative of your program.

### 2.3 Function Definition

There are two ways to define functions:

```bash
function name {
  commands
  return
}

name () {
  commands
  return
}
```

Spaces between curly braces and commands are required!

Shell function definitions must appear in the script before they are called.

### 2.4 Function Parameters

Passing parameters to a function is similar to shell script parameters.

**example.sh:**

```bash
#!/bin/bash
echo $1
function func {
   echo $1
}
func "hello function"
$ chmod u+x ./example.sh && ./example.sh "hello script"

hello function
```

### 2.5 Return Parameters

The variable “$?” contains the return status of a function.

The keyword return is used to return the status.

If the keyword is not specified and the function executes successfully, “$?” will be 0.

Example:
```bash
#!/bin/bash

function func {
   return 1
}

func
echo "Return code is $?"
```

What is the return code?



````python
#What is the return code of the above Bash code?
#1
````

### 2.6 Scoping in Bash

Bash uses dynamic scoping.

By default, variables are declared globally, meaning that they can be accessed and modified from anywhere in the script.

Local variables are defined only for the context in which it was created.

```bash
#!/bin/bash

VAR="global variable"

function func {
   local VAR="local variable"
   echo $VAR
}

func
echo $VAR
```



````python
#What is the output of above Bash code?
#local variable
#global variable
````
